```python
# Project Calculator 
import math
class Calculator:
    def __init__(self):
        self.history = []

    #History
    def _record(self,expression,result):
        self.history.append(f"{expression} = {result}")

    def show_history(self):
        if not self.history:
            print("No history yet.")
            return
        print("\n── Calculation History ──")
        for i, entry in enumerate(self.history, 1):
            print(f"{i}. {entry}")
        print()
 
    def clear_history(self):
        self.history.clear()

        
  #Addition 
    def add(self,a,b):
        result = a +b 
        self._record(f"{a} +{b}", result)
        return result 

    # Subtraction
    def sub(self,a,b):
        result = a - b 
        self._record(f"{a}-{b}", result)
        return result 
    
    # Multiplication
    def mutliply(self,a,b):
        result = a*b
        self._record(f"{a}*{b}", result)
        return result
        
    # Division
    def division(self,a,b):
        try:
            result = a/b
            self._record(f"{a}/{b}", result)
            return result
        except ZeroDivisionError:
            print("Cannot divide by zero")
        except TypeError:
            print("Inputs must be numbers")

            

    # to find MOdulo or Remainder of the number

    def modulo(self, a, b):
        if b == 0:
            raise ValueError("Cannot modulo by zero")
        result = a % b
        self._record(f"{a} % {b}", result)
        return result


        

     # Logarithm Function 
    
    def log(self, a, base=10):
        if a <= 0:
            raise ValueError("Logarithm input must be positive")
        result = math.log(a, base)
        self._record(f"log{base}({a})", result)
        return result
        
 
    def ln(self, a):
        if a <= 0:
            raise ValueError("Natural log input must be positive")
        result = math.log(a)
        self._record(f"ln({a})", result)
        return result

    # Absolute Value
    def absolute(self, a):
        result = abs(a)
        self._record(f"|{a}|", result)
        return result


    
    #Power of any number
    def power(self, base, exp):
        result = base ** exp
        self._record(f"{base} ^ {exp}", result)
        return result

    # Square root 
    def square_root(self,a):
        try:
            result = a**0.5
            self._record(f"{a}^0.5",result)
        except ValueError:
            return "Input must be positive"
    # Pi 
    def pi_value(self,a):
        result = a * math.pi
        self._record(f"{a}*3.14",result)
        return result

    # Factorial 

    def fac(self,num):
         if not isinstance(num, int) or num < 0:
             raise ValueError("Factorial requires a non-negative integer")
         if num > 20:
            raise ValueError("Input too large for factorial (max 20)")
         result = math.factorial(num)
         self._record(f"{num}!", result)
         return result

    # Inverse of the number 

    def inverse(self, a):
        if a == 0:
            raise ValueError("Cannot take inverse of zero")
        result = 1 / a
        self._record(f"1/{a}", result)
        return result

    
    # Trigonoemtric Function

    def sin(self, degrees):
        result = math.sin(math.radians(degrees))
        self._record(f"sin({degrees}°)", result)
        return result
  
    def cos(self, degrees):
        result = math.cos(math.radians(degrees))
        self._record(f"cos({degrees}°)", result)
        return result

    def tan(self, degrees):
        if degrees % 180 == 90:
            raise ValueError("tan is undefined at 90° and 270°")
        result = math.tan(math.radians(degrees))
        self._record(f"tan({degrees}°)", result)
        return result


calc = Calculator()
print(calc.add(10, 20))
print(calc.sub(40,30))
calc.division(40,0)
```
